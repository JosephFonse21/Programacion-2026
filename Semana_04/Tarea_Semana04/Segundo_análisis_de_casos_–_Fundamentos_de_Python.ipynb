# Problema #1: Calculadora de propina inteligente

## Problema

En un restaurante se solicita hacer un calculo total de la cuenta a pagar con la propina inlcluida.

## Entrada
Variable factura donde se va a digitar el monto de la factura.
Variable propina donde se va a digitar el monto de propina
Variable calculo_propina donde se va a hacer el el calculo del monto de la propina.
y Variable total donde se va almacenar el monto de la factura mas el monto de la propina
-

## Proceso
calculo_propina = factura * (propina / 100) // este procedimiento va a calcular la propina del cliente, donde el usuario va escribir la cantidad de propina en porcentaje es decir por ejemplo 10 y esa cantidad se divide entre 100 para poder hacer el calculo correcto de la propina.

total = factura + calculo_propina // este procedimiento es para obtner el resultado total, la cuenta a pagar mas el porcentaje de propina.
-

## Salida esperada
print(f'La cantidad de propina es: {calculo_propina} ')// donde se imprime la cantida de propina que se quiere pagar.

print(f'El total a pagar es: {total}') // Y el total a pagar.
-

```python
# Escriba la solución aquí

#Entradas
try:
    factura = float(input('Cuanto es el monto de la factura: '))
except ValueError:
    print('Error monto de dinero invalido, asignado -1')
    factura = -1

try:
    propina = float(input('Cual es el monto en porcentaje que desea agregar: '))
except ValueError:
    print('Error monto de propina invalido, asignado -1')
    propina = -1

calculo_propina = 0
total = 0


#Proceso

calculo_propina = factura * (propina / 100)
total = factura + calculo_propina

#Salida
print(f'La cantidad de propina es: {calculo_propina} ')
print(f'El total a pagar es: {total}')
```

# Problema #2: Detector de número par/impar

## Problema

Se quiere hacer un programa en el que se le solicite al usuario un numero para ver si el numero digitado es par o impar

## Entrada
Variable numero que va a funcionar como entrada para que el usuario digite un numero entero.Si digita un numero que no es entero la variable numero se convierte en -1

## Proceso y Salida esperada
si el numero digitado es menor que 0 se imprime un error de que el numero digitado no es entero.
Sino si el numero digitado se divide entre 2 y si su residuo es 0 entonces imprime que el numero es par.
Sino imprime que el numero digitado es impar.


```python
# Escriba la solución aquí
#Entrada
try:
    numero = int(input('Digite un numero entero para ver si es par o impar: '))
except ValueError:
    numero = -1


#Proceso y salida
if numero < 0:
    print('Error, numero digitado no es entero!')
elif numero % 2 == 0:
    print('El numero digitado es Par !! ')
else:
    print('El numero digitado es Impar !! ')
```

# Problema #3: Clasificador de edad

## Problema

Se solicita un programa que clasifique en categorias dependiendo la edad que tenga el usuario

## Entrada
Una variable edad que solicita la edad al usuario si escribe un numero negativo va a dar error.
-

## Proceso y Salida esperada
Si la edad es menor que 0 se imprime que la edad digitada es invalida.
Sino si la edad es mayor o igual a 0 pero menor que 18 entonces imprime que es menor de edad.
Sino si la edad es mayor o igual a 18 pero menor que 65 entonces imprime que es un Adulto.
Si no imprime que es un adulto mayor.



```python
# Escriba la solución aquí

#Entrada
try:
    edad = int(input('Digite su edad: '))
except ValueError:
    numero = -1


#Proceso y salida
if edad < 0:
    print('Edad digitada es invalida')
elif edad >= 0 and edad < 18:
    print('Eres menor de edad')
elif edad >= 18 and edad < 65:
    print('Eres un Adulto')
else:
    print('Eres un adulto mayor')
```

# Problema #4: Validador de compra con descuento

## Problema
Se solicita un programa que dependiendo de las entradas que digite el usuario se le hace descuento a la compra que realizo o no.

## Entrada
Una variable cliente que solicite el nombre del cliente.
Una variable monto que solicite el monto de la compra del cliente.
Una variable cliente_frecuente que solicite al usuario que escriba si es un cliente frecuente o no.
Una variable calculo_descuento en el que se va a hacer el calculo del descuento.
Una variable total en el que se le va a asignar el total de la compra.
-

## Proceso y Salida esperada
si el monto es mayor a 5000 y es cliente frecuente entonces se le aplica un 10% de descuento al monto de la compra, imprime de cuanto va a ser el descuento y el total a pagar.
Sino imprime el total a pagar sin el descuento.



```python
# Escriba la solución aquí
#Entrada
cliente = input('Digite su nombre: ').lower()
try:
    monto = float(input('Digite el monto de la compra: '))
except ValueError:
    print('Monto invalido, asignado -1 ')
    monto = -1
cliente_frecuente = input('Eres cliente frecuente?: si/no ').lower().strip() == 'si'
calculo_descuento = 0
total = 0

#Proceso y salida

if cliente_frecuente and monto > 5000 :
    calculo_descuento = monto * 0.1
    total = monto - calculo_descuento
    print(f'{cliente} por ser cliente frecuente tienes un 10% de descuento que es de {calculo_descuento}')
    print(f'Entonces el total de la compra es de: {total}')

else:
    total = monto
    print(f'{cliente} el monto total es: {total} ')
```
